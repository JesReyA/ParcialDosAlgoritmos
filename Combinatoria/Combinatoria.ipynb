```python
# Obtener combinación de pasos para llegar de un punto a otro dentro del plano cartesiano

def obtener_pasos(inicio, fin):
    x_1, y_1 = inicio
    x_2, y_2 = fin

    dx = x_2 - x_1
    dy = y_2 - y_1

    pasos = []

    pasos.extend(["R"]*dx)
    pasos.extend(["U"]*dy)

    return pasos
```

```python
print(obtener_pasos((1,1), (6,4)))
```

```python
# Combinaciones sin librería

total_combinaciones = 0
def combinaciones_algoritmo(lista, cantidad_elementos):
    global total_combinaciones
    total_combinaciones=0
    generar_combinaciones ([], sorted(lista), cantidad_elementos)
    print(f"{total_combinaciones}")

def generar_combinaciones(combinacion_actual, opciones, cantidad_elementos):
    if(len(combinacion_actual) == cantidad_elementos):
        print(combinacion_actual)
        global total_combinaciones
        total_combinaciones +=1
        return

    copia_opciones = opciones.copy()
        
    for elemento in sorted(set(opciones)):
        nueva_combinacion = combinacion_actual + [elemento]
        copia_opciones.remove(elemento)
        generar_combinaciones(nueva_combinacion, copia_opciones.copy(), cantidad_elementos)

        while elemento in copia_opciones:
            copia_opciones.remove(elemento)
```

```python
lista = ["R","R","R","R","R","U","U","U"]
print(combinaciones_algoritmo(lista ,8))
```

```python
# Combinaciones con librería

from itertools import combinations

def combinaciones_libreria(lista, elementos):
    resultado = set(combinations(lista, elementos))
    print(resultado)
```

```python
lista = ["R","R","R","R","R","U","U","U"]
print(combinaciones_libreria(lista ,5))
```

```python
# Permutaciones sin librería

total_permutaciones = 0
def permutaciones_algoritmo(lista):
    global total_permutaciones
    total_permutaciones=0
    generar_permutaciones ([], lista)
    print(f"{total_permutaciones}")

def generar_permutaciones(permutacion_actual, opciones):
    if(len(opciones) == 0):
        print(permutacion_actual)
        global total_permutaciones
        total_permutaciones +=1
        return
        
    for elemento in set(opciones):
        nueva_permutacion = permutacion_actual + [elemento]
        copia_opciones = opciones.copy()
        copia_opciones.remove(elemento)
        generar_permutaciones(nueva_permutacion, copia_opciones)
```

```python
lista = ["R","R","R","R","R","U","U","U"]
permutaciones_algoritmo(lista)
```

```python
# Permutaciones con librería

from itertools import permutations

def permutaciones(lista):
    resultado = set(permutations(lista))
    return list(resultado)
```

```python
lista = ["R","R","R","R","R","U","U","U"]
print(permutaciones(lista))
print(len(permutaciones(lista)))
```
