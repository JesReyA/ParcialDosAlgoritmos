```python
import pandas as pd
import networkx as nx
from IPython.display import HTML
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
DG = nx.Graph()
```

```python
datos=pd.read_csv('mapeo_completo.csv')
datos.head()
```

```python
numeracion_estaciones = pd.read_csv('numeracion_estaciones.csv')
numeracion_estaciones.head()
```

```python
for filas in datos.iterrows():
    DG.add_edge(filas[1]["origen"], filas[1]["destino"],peso= filas[1]["peso"]);
DG.nodes(data=True)
```

```python
nx.draw_circular(
    DG,
    node_color='lightblue',
    edge_color='gray',
    font_size=10,
    width=4,
    with_labels=True,
    node_size=350,
)
```
